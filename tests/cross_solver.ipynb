```python
from solver import Solver
from magiccube import Cube
```

```python
i = 0
solved = 0
while i < 500:
    cube = Cube()
    cube.rotate("x2")
    cube.scramble(num_steps=15)
    history = cube.history(to_str=True)

    solver = Solver(
        env="cross",
        cube = cube
    )

    solution = " ".join(solver.solve())
    done = solver.env.is_terminated(solver.cube)

    if not done:
        print(f"scramble {i + 1} not solved: {solution}")
        print(f"state: {cube.get()} -> '{history}' ")
        #print(cube)
        #print(solver.cube)
    else:
        solved += 1
        #print(f"scramble {i + 1} solved: {solution}")
        #print(solver.cube)
        pass
    print("-" * 50)
    i += 1

print(f"status: {solved}/500")
```
