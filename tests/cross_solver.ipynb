```python
from solver import Solver
from magiccube import Cube
```

```python
i = 0
while i < 500:
    cube = Cube()
    cube.rotate("x2")
    cube.scramble()

    solver = Solver(
        env="cross",
        cube = cube
    )

    solution = " ".join(solver.solve())
    done = solver.env.is_terminated(solver.cube)

    if not done:
        print(cube)
        print(solver.cube)
        print("-" * 50)
    else:
        print(f"scramble {i + 1} solved: {solution}")
        print(solver.cube)
    i += 1
```
